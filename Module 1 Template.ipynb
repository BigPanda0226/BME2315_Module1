1. Prevalence and incidence:

    Prevalence is the number of people living with a disease, while incidence is the number of new cases that develop over a period of time. In 2026, an estimated 7.4 million Americans age 65 and older are living with Alzheimer’s dementia, or about 1 in 9 people in that age group. Incidence rises sharply with age: a study cited in the 2026 Alzheimer’s Association report estimated annual rates of 0.4% at ages 65–74, 3.2% at ages 75–84, and 7.6% at age 85 or older. Those incidence rates come from 2011 data and illustrate the age pattern rather than the current number of new cases.

        https://www.alz.org/alzheimers-dementia/facts-figures

2. Economic burden:

    Alzheimer’s disease creates costs for patients, families, insurers, and care systems. In 2026, U.S. health and long-term care spending for people with Alzheimer’s or dementia is projected to reach $409 billion. Medicare and Medicaid are expected to pay $263 billion of that amount, while patients and families are expected to pay $103 billion out of pocket. The burden also extends beyond those payments. In 2025, unpaid caregivers provided more than 19 billion hours of care, valued at over $446 billion.

        https://www.alz.org/alzheimers-dementia/facts-figures

3. Risk factors (genetic and lifestyle):

    Older age is the strongest known risk factor for Alzheimer’s disease, but genetics also play a significant role. Studies show carrying the APOE ε4 variant increases risk, although it does not mean a person will definitely develop Alzheimer’s. Cardiovascular and metabolic conditions, including high blood pressure and diabetes, are associated with dementia risk. Physical activity, avoiding smoking, and managing those health conditions are ways people may reduce their risk, though no lifestyle choice guarantees prevention.

        https://www.cdc.gov/alzheimers-dementia/prevention/index.html

4. Societal determinants:

    Social conditions can affect both dementia risk and a person’s ability to receive care. The CDC identifies education, access to health care, the physical environment, and social connection as factors that may influence dementia risk. Limited access to health care can also delay recognition of symptoms and diagnosis, which may delay care planning and support. These factors help explain why Alzheimer’s affects communities differently and why improving access to care matters alongside studying biological risk.

        https://www.cdc.gov/alzheimers-dementia/php/sdoh/index.html

5. Symptoms:

    Memory problems, especially difficulty remembering recent information or short-term memory, are often among the first noticeable symptoms of Alzheimer’s disease. As the disease progresses, people may have trouble thinking clearly, finding words, making decisions, or completing familiar tasks. Changes in mood, behavior, or personality can also occur. Symptoms generally worsen over time and can eventually interfere with everyday activities and independence.

        https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447

6. Diagnosis:

    A physician begins by discussing a person’s symptoms and medical history, often asking someone close to the patient about changes they have noticed. Cognitive testing helps assess memory and thinking, while a physical examination, laboratory tests, and brain imaging may help rule out other causes of symptoms. In some cases, tests for Alzheimer’s-related biomarkers in cerebrospinal fluid, blood, or brain scans can provide additional evidence. Diagnosis brings these findings together.

        https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers/art-20048075
