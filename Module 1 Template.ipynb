10. Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

    In healthy neurons, dendrites receive signals from other neurons, while electrical signals travel down the axon, which is insulated by the myelin sheath, to axon terminals that release neurotransmitters across the synapse. In Alzheimer’s disease, abnormal accumulation of amyloid-beta 40 and 42, particularly Aβ42, contributes to plaques around neurons and disrupts synaptic communication. Tau normally stabilizes microtubules that help transport materials along the axon, but in Alzheimer’s, abnormal tau forms neurofibrillary tangles, disrupting transport and contributing to neuronal dysfunction and death. As neurons and their connections are progressively damaged, brain regions involved in memory and cognition begin to deteriorate.

## Data Analysis: 

1. We used Python to investigate whether sex, age at death, and known head injury history were associated with Aβ42 levels in the donor dataset. We loaded the CSV file into our Patient class, which stored each donor’s demographic information, cognitive status, head injury history, and protein measurements. We also sorted the donors by age at death.
    
        from patient_cherry import Patient
        import matplotlib.pyplot as plt
        import numpy as np
        import statistics
        from scipy import stats
        import pandas as pd
        from sklearn.linear_model import LinearRegression

        filename = "Metadata and Protein Data for Module 1.csv"
    
        Patient.instantiate_from_csv(filename)

        Patient.all_patients.sort(
            key=Patient.get_age_at_death,
            reverse=False
        )

        print("Number of patients loaded:", len(Patient.all_patients))
        print("Unique sex values:", set(repr(p.sex) for p in Patient.all_patients))
        print("Unique cognitive status values:", set(repr(p.cognitive_status) for p in Patient.all_patients))

2. For the sex comparison, we selected donors with dementia and separated them into female and male groups. We collected their Aβ42 concentrations and calculated the mean and standard deviation for each group.

        male_dementia_patients = Patient.filter(
            Patient.all_patients,
            sex="Male",
            cognitive_status="Dementia"
        )

        female_abeta42 = []
        male_abeta42 = []

        for patient in female_dementia_patients:
            female_abeta42.append(patient.abeta42)

        for patient in male_dementia_patients:
            male_abeta42.append(patient.abeta42)

        female_mean = statistics.mean(female_abeta42)
        male_mean = statistics.mean(male_abeta42)
        female_stdev = statistics.stdev(female_abeta42)
        male_stdev = statistics.stdev(male_abeta42)

3. We used a two-sided, unpaired t test because we were comparing two separate groups and asking whether their means differed in either direction. This was not a repeated measures test because each donor contributed one Aβ42 measurement to the comparison. We used a significance threshold of 0.05. Our stats.ttest_ind call uses the default equal-variance assumption.

        t_stat, p_value = stats.ttest_ind(
            female_abeta42,
            male_abeta42
        )

        print(f"t-statistic: {t_stat:.2f}")
        print(f"p-value: {p_value:.4f}")

4. We displayed the group means in a bar graph and added the t statistic and p value. Our code draws error bars extending one standard deviation above each mean.

        sex_labels = ["Female", "Male"]
        abeta42_means = [female_mean, male_mean]

5. For the age analysis, we collected each donor’s age at death and Aβ42 concentration. We calculated absolute z scores for the Aβ42 values and excluded values more than three standard deviations from the mean. This removed one donor who died at age 81 and had an Aβ42 concentration of 1412 pg/µg. The exclusion applied only to the age analysis; it did not remove the donor from Patient.all_patients or the bar graph comparisons.

        ages_at_death = []
        all_abeta42 = []

        for patient in Patient.all_patients:
            ages_at_death.append(patient.age_at_death)
            all_abeta42.append(patient.abeta42)

        all_abeta42_array = np.array(all_abeta42)
        z_scores = np.abs(stats.zscore(all_abeta42_array))

        outlier_mask = z_scores > 3
        outlier_indices = np.where(outlier_mask)[0]

        print(
            f"\nNumber of outliers detected (|z| > 3): "
            f"{len(outlier_indices)}"
        )

        for i in outlier_indices:
            print(
                f"Outlier: {Patient.all_patients[i].donor_id}, "
                f"Age at Death = {ages_at_death[i]}, "
                f"ABeta42 = {all_abeta42[i]:.2f} pg/ug, "
                f"z = {z_scores[i]:.2f}"
            )

        ages_at_death_clean = [
            age for i, age in enumerate(ages_at_death)
            if not outlier_mask[i]
        ]

        all_abeta42_clean = [
            val for i, val in enumerate(all_abeta42)
            if not outlier_mask[i]
        ]

6. We performed simple linear regression on the remaining observations, using age at death as the independent variable and Aβ42 as the dependent variable. We calculated the slope, intercept, and R² value. R² describes the proportion of variation in Aβ42 explained by age in this linear model.

        ages_at_death_array = np.array(
            ages_at_death_clean
        ).reshape(-1, 1)

        model = LinearRegression()
        model.fit(ages_at_death_array, all_abeta42_clean)

        predicted_abeta42 = model.predict(ages_at_death_array)

        slope = model.coef_[0]
        intercept = model.intercept_
        r_squared = model.score(
            ages_at_death_array,
            all_abeta42_clean
        )

        print(f"\nRegression slope (outliers removed): {slope:.2f}")
        print(f"Regression intercept (outliers removed): {intercept:.2f}")
        print(f"R-squared value (outliers removed): {r_squared:.4f}")

7. We plotted the remaining donor measurements as individual points and added the fitted regression line. We displayed the equation and R² on the scatterplot and saved the figure as a PNG file.

        plt.figure(figsize=(7, 5))

        plt.scatter(
            ages_at_death_clean,

8. For the head injury comparison, we separated donors whose known head injury history was recorded as “Yes” or “No.” Donors with missing head injury information were excluded from this comparison. We calculated the mean and standard deviation of Aβ42 for each group and performed another two-sided, unpaired t test using the same significance threshold of 0.05.

        head_injury_yes = Patient.filter(

            head_injury="Yes"
        )

        head_injury_no = Patient.filter(

        yes_abeta42 = [patient.abeta42 for patient in head_injury_yes]
        no_abeta42 = [patient.abeta42 for patient in head_injury_no]

        yes_mean = statistics.mean(yes_abeta42)
        no_mean = statistics.mean(no_abeta42)
        yes_stdev = statistics.stdev(yes_abeta42)
        no_stdev = statistics.stdev(no_abeta42)

        print(f"p-value: {injury_p_value:.4f}")

## Conclusions and Ethical Implications: 
Female donors with dementia had a higher mean Aβ42 concentration than male donors with dementia, but this difference was not statistically significant (p = 0.4577). Donors with a known head injury had a lower mean Aβ42 concentration than donors without a known head injury, but this difference was also not statistically significant (p = 0.1328). After excluding one outlier, the age regression produced an R² of 0.0108, meaning that age at death explained approximately 1.1% of the variation in Aβ42 concentrations. These findings did not provide strong evidence of differences by sex or head injury history, or of a strong linear relationship with age. They also do not prove that the groups are identical or that no relationship exists.

These uncertain findings should not be used to predict an individual patient’s Aβ42 level or Alzheimer’s risk based only on sex or head injury history. Presenting the observed group differences as established facts could create unnecessary worry or lead to unfair assumptions about patients. We should report the nonsignificant results clearly and disclose the outlier exclusion so readers can understand how we reached our conclusions.

## Limitations and Future Work: 
Our analysis was limited by the sample size and the large variation in Aβ42 concentrations. Head injury history was recorded only as “Yes” or “No,” so we could not examine injury severity, timing, or the number of injuries. Missing injury histories also reduced the number of donors available for that comparison. We examined each factor separately without accounting for other variables, such as APOE genotype or cognitive status, that might help explain differences between donors. Additionally, the shared code did not include tests of normality or equal variance, which are relevant assumptions for the t tests we used. Because the dataset was observational, our results cannot establish causation.

Future work could compare the age regression with and without the excluded donor to determine how much that observation affects the result. We could also study more donors and compare groups with similar ages and cognitive status. Examining Aβ40, tau measurements, and APOE genotype alongside Aβ42 could provide a more complete explanation of the variation between donors. A separate study that follows living participants over time could investigate changes after head injury using a biomarker that can be measured repeatedly, since this dataset contains measurements collected from donors at death.
